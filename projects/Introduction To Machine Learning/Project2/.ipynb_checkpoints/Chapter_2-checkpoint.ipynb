````python
plt.figure(figsize=(6, 4))
plt.scatter(train['x0'], train['x1'], c=train['y'], cmap='coolwarm')
plt.grid(True)
plt.xlabel('x0', fontsize=14)
plt.ylabel('x1', fontsize=14)
plt.title('x1 vs. x0', fontsize=16)
plt.colorbar()
plt.show()
````

````python
import torch
import torch.nn as nn
import torch.optim as optim
from torch.utils.data import DataLoader, TensorDataset
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import accuracy_score
````

````python
# Setting the seed for torch and np as 42
torch.manual_seed(42)
np.random.seed(42)
````

````python
X_train = train[['x0', 'x1']].values
y_train = train['y'].values
X_test = test[['x0', 'x1']].values
y_test = test['y'].values
````

````python
#Normalize features
scaler = StandardScaler()
X_train = scaler.fit_transform(X_train)
X_test = scaler.transform(X_test)
````

````python
# Convert to tensors
X_train_t = torch.tensor(X_train, dtype=torch.float32)
y_train_t = torch.tensor(y_train, dtype=torch.float32).view(-1, 1)
X_test_t = torch.tensor(X_test, dtype=torch.float32)
y_test_t = torch.tensor(y_test, dtype=torch.float32).view(-1, 1)

train_ds = TensorDataset(X_train_t, y_train_t)
train_dl = DataLoader(train_ds, batch_size=32, shuffle=True)
````

````python
class MLP(nn.Module):
    def __init__(self):
        super().__init__()
        self.net = nn.Sequential(
            nn.Linear(2, 64),
            nn.ReLU(),
            nn.Linear(64, 32),
            nn.ReLU(),
            nn.Linear(32, 16),
            nn.ReLU(),
            nn.Linear(16, 1),
            nn.Sigmoid()
        )

    def forward(self, x):
        return self.net(x)

model = MLP()
````

````python
# Optimizer and loss
optimizer = optim.Adam(model.parameters(), lr=0.01)
criterion = nn.BCELoss()
````

````python
# LR Scheduler - warmup and cosine Decay
def lr_schedule(step, warmup_steps=20, total_steps=200):
    if step < warmup_steps:
        return step / warmup_steps
    progress = (step - warmup_steps) / (total_steps - warmup_steps)
    return 0.5 * (1 + np.cos(np.pi * progress))  # cosine decay
````

````python
# Loop
epochs = 100
train_accs, test_accs, lrs = [], [], []

for epoch in range(epochs):
    model.train()
    for xb, yb in train_dl:
        optimizer.zero_grad()
        preds = model(xb)
        loss = criterion(preds, yb)
        loss.backward()
        optimizer.step()

    lr = 0.01 * lr_schedule(epoch)
    for param_group in optimizer.param_groups:
        param_group['lr'] = lr
    lrs.append(lr)

    # Eval
    model.eval()
    with torch.no_grad():
        y_pred_test = (model(X_test_t) > 0.5).float()
        acc = accuracy_score(y_test, y_pred_test)
        test_accs.append(acc)
        y_pred_train = (model(X_train_t) > 0.5).float()
        train_accs.append(accuracy_score(y_train, y_pred_train))

    print(f"Epoch {epoch+1:03d} | LR: {lr:.5f} | Test Acc: {acc:.4f}")
    #if acc >= 0.93:
    #    print(f"Ended early")
     #   break
````

````python
print(f"Final Test Accuracy: {test_accs[-1]:.4f}")
````

**3.5 Conduct a simple ablation study of each architectural and optimization choice concerning test accuracy.**
I have found all architectural or optimization parts that affect the accuracy and its as follows:
|Category	|Choice	|Ablation idea|
|---------------|---------------|-------|
Learning rate	|0.01	| Try fixed LR vs. scheduled LR|
LR scheduler|	Warmup + Cosine decay|	Remove scheduler (constant LR)|
Optimizer|	Adam	|Replace with SGD|
Activation|	ReLU	|Replace with Tanh|
Architecture	|3 hidden layers (64→32→16)	|Try fewer layers|
Batch size|	32	|Try full batch size |

Under here I will use my ablation idea and see the result of replacing the part with the ablation idea.

````python
def train_and_evaluate(model, optimizer_fn, lr=0.01, use_scheduler=True, epochs=100, batch_size=32):
    train_dl = DataLoader(train_ds, batch_size=batch_size, shuffle=True)
    criterion = nn.BCELoss()
    optimizer = optimizer_fn(model.parameters(), lr=lr)
    train_accs, test_accs = [], []

    for epoch in range(epochs):
        model.train()
        for xb, yb in train_dl:
            optimizer.zero_grad()
            preds = model(xb)
            loss = criterion(preds, yb)
            loss.backward()
            optimizer.step()

        # Optional LR scheduler
        if use_scheduler:
            lr_epoch = lr * lr_schedule(epoch)
            for param_group in optimizer.param_groups:
                param_group['lr'] = lr_epoch

        model.eval()
        with torch.no_grad():
            y_pred_test = (model(X_test_t) > 0.5).float()
            test_acc = accuracy_score(y_test, y_pred_test)
            y_pred_train = (model(X_train_t) > 0.5).float()
            train_acc = accuracy_score(y_train, y_pred_train)
            test_accs.append(test_acc)
            train_accs.append(train_acc)
    
    return max(test_accs)
````

````python
results = {}

# 1. Baseline
results['Baseline (ReLU, Adam, CosineLR)'] = train_and_evaluate(MLP(), optim.Adam)

# 2. No LR scheduler
results['No LR scheduler'] = train_and_evaluate(MLP(), optim.Adam, use_scheduler=False)

# 3. SGD
results['SGD optimizer'] = train_and_evaluate(MLP(), optim.SGD)

# 4. Tanh
class MLP_Tanh(nn.Module):
    def __init__(self):
        super().__init__()
        self.net = nn.Sequential(
            nn.Linear(2, 64),
            nn.Tanh(),
            nn.Linear(64, 32),
            nn.Tanh(),
            nn.Linear(32, 16),
            nn.Tanh(),
            nn.Linear(16, 1),
            nn.Sigmoid()
        )
    def forward(self, x):
        return self.net(x)
results['Tanh activations'] = train_and_evaluate(MLP_Tanh(), optim.Adam)

# 5. Fewer layers
class MLP_Simple(nn.Module):
    def __init__(self):
        super().__init__()
        self.net = nn.Sequential(
            nn.Linear(2, 32),
            nn.ReLU(),
            nn.Linear(32, 1),
            nn.Sigmoid()
        )
    def forward(self, x):
        return self.net(x)
results['Shallower network'] = train_and_evaluate(MLP_Simple(), optim.Adam)

# 6. Batch size 
results['Full batch (batch_size = all)'] = train_and_evaluate(
    MLP(), optim.Adam, batch_size=len(train_ds)
)
````

## 🎯 Note: Clustering Accuracy Evaluation

The clustering accuracy metric evaluates how well the predicted clusters match the true clusters, irrespective of the specific labels or names assigned to the clusters.

This means that the evaluation is solely based on the correct grouping of data points rather than the numerical labels of the clusters themselves.

## Key Characteristics

- **Name-Independent**: The metric cares only about how well the data points are grouped together, not the specific numerical or categorical labels used for the clusters.
- **Focus on Grouping**: The evaluation rewards cluster assignments that correctly group the same data points together, regardless of the specific labels assigned.

### Example

If the true cluster assignments are:

```
[0, 0, 0, 1, 1, 1]
```

and your predicted cluster assignments are:

```
[1, 1, 1, 0, 0, 0]
```

the accuracy will still be **1.0** because the grouping of points is identical, even though the numerical labels are swapped.

## How the Metric is Computed

1. **Contingency Matrix**: 
   - Construct a contingency matrix that represents the overlap between the true clusters and the predicted clusters.

2. **Optimal Correspondence**: 
   - Use an optimization algorithm, such as the Hungarian method (linear sum assignment), to find the best possible correspondence between true and predicted labels, maximizing the number of correctly assigned data points.

3. **Accuracy Calculation**: 
   - Calculate the accuracy as the ratio of correctly matched data points to the total number of data points.

> This approach ensures that the evaluation is based on **cluster completeness** and **homogeneity**, rewarding cluster assignments that correctly group the same data points together, regardless of the specific labels used.

---

````python
import numpy as np
import scipy.optimize
import sklearn.metrics

def calculate_clustering_accuracy(true_labels: np.ndarray, cluster_labels: np.ndarray) -> float:
    """
    Calculate the clustering accuracy between true labels and cluster labels.

    The function constructs a contingency matrix between the true labels and cluster labels.
    It then uses the Hungarian algorithm (also known as linear sum assignment) to find the
    best matching between the cluster labels and true labels. The clustering accuracy is 
    calculated as the number of correctly matched labels divided by the total number of labels.

    Args:
        true_labels (np.ndarray): An array of true labels for each data point.
        cluster_labels (np.ndarray): An array of cluster labels predicted by a clustering algorithm.

    Returns:
        float: The clustering accuracy, a value between 0 and 1 representing the proportion of 
               correctly matched labels.
    
    Example:
        >>> true_labels = np.array([0, 1, 2, 0, 1, 2])
        >>> cluster_labels = np.array([1, 2, 0, 1, 2, 0])
        >>> calculate_clustering_accuracy(true_labels, cluster_labels)
        1.0

    Raises:
        ValueError: If true_labels and cluster_labels are not of the same length.
    """
    # Check if the input labels are of the same length
    if true_labels.size != cluster_labels.size:
        raise ValueError("true_labels and cluster_labels must have the same length.")

    # Construct a contingency matrix where each cell [i, j] indicates the number of points with 
    # true label i and cluster label j.
    contingency_matrix = sklearn.metrics.cluster.contingency_matrix(true_labels, cluster_labels)
    
    # Find the best matching between true labels and cluster labels using the Hungarian algorithm.
    # We negate the contingency matrix because linear_sum_assignment finds the minimum cost assignment.
    row_ind, col_ind = scipy.optimize.linear_sum_assignment(-contingency_matrix)
    
    # Calculate the number of correctly assigned labels according to the optimal matching.
    correct_pairs = contingency_matrix[row_ind, col_ind].sum()
    
    # Compute the clustering accuracy as the ratio of correctly matched labels to total labels.
    accuracy = correct_pairs / true_labels.size

    return accuracy
````

````python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.preprocessing import MinMaxScaler
from sklearn.cluster import DBSCAN
import umap.umap_ as umap
from scipy.optimize import linear_sum_assignment
from sklearn.metrics import cluster

# === Load data ===
data = pd.read_csv("unsupervised_data.csv")
data.reset_index(drop=False, inplace=True)
data.rename(columns={'index': 'ID'}, inplace=True)
X = data.drop(columns=['ID']).values

num_samples = X.shape[0]
X_images = X.reshape(num_samples, 1, 128, 128)

# Step 2: Optional: extract features per image
# For DBSCAN/KMeans, flatten again
X_features = X_images.reshape(num_samples, -1)  # back to (num_samples, 16384)


# === Scale ===
scaler = MinMaxScaler()
X_scaled = scaler.fit_transform(X_features)

# === UMAP dimensionality reduction ===
reducer = umap.UMAP(
    n_neighbors=19,       # slightly higher than before
    min_dist=0.1,        # keeps clusters tight but separate
    n_components=4,       # 3D space sometimes gives DBSCAN more structure
    metric='cosine',
    random_state=42
)
X_umap = reducer.fit_transform(X_scaled)
print("UMAP shape:", X_umap.shape)

# === Visualize the 2D projection ===
plt.figure(figsize=(6,5))
plt.scatter(X_umap[:,0], X_umap[:,1], s=8, alpha=0.6)
plt.title("UMAP projection (unsupervised_data)")
plt.xlabel("UMAP-1")
plt.ylabel("UMAP-2")
plt.show()


````

````python

dbscan = DBSCAN(eps=4, min_samples=10)  
clusters = dbscan.fit_predict(X_umap)

print("DBSCAN unique clusters:", np.unique(clusters))
plt.figure(figsize=(6,5))
plt.scatter(X_umap[:,0], X_umap[:,1], c=clusters, cmap='tab10', s=8)
plt.title("DBSCAN Clustering on UMAP space")
plt.xlabel("UMAP-1")
plt.ylabel("UMAP-2")
plt.show()

# === Optional: remove noise cluster (-1) if present ===
noise_mask = clusters != -1
final_clusters = clusters.copy()
final_clusters[~noise_mask] = np.max(clusters[noise_mask]) + 1  # assign new cluster for noise if any

# === Save submission ===
submission_path = "submission_umap_dbscan.csv"
pd.DataFrame({'ID': data['ID'], 'cluster': final_clusters}).to_csv(submission_path, index=False)
print("Saved:", submission_path)

# === Evaluate locally (if correct_submission.csv available) ===
correct = pd.read_csv("correct_submission.csv")
merged = pd.merge(pd.DataFrame({'ID': data['ID'], 'cluster_pred': final_clusters}), correct, on='ID')

def clustering_accuracy(true_labels, cluster_labels):
    contingency = cluster.contingency_matrix(true_labels, cluster_labels)
    row_ind, col_ind = linear_sum_assignment(-contingency)
    return contingency[row_ind, col_ind].sum() / true_labels.size

acc_hungarian = clustering_accuracy(merged['cluster'], merged['cluster_pred'])
print(f"Hungarian-matched clustering accuracy: {acc_hungarian:.4f}")
````

````python
import numpy as np
import pandas as pd
from sklearn.preprocessing import MinMaxScaler
from sklearn.cluster import DBSCAN
import umap.umap_ as umap
from scipy.optimize import linear_sum_assignment
from sklearn.metrics import cluster
import matplotlib.pyplot as plt
from itertools import product

# === Load data ===
data = pd.read_csv("unsupervised_data.csv")
data.reset_index(drop=False, inplace=True)
data.rename(columns={'index': 'ID'}, inplace=True)
X = data.drop(columns=['ID']).values

num_samples = X.shape[0]
X_images = X.reshape(num_samples, 1, 128, 128)
X_features = X_images.reshape(num_samples, -1)  # flatten for clustering

# === Scale features ===
scaler = MinMaxScaler()
X_scaled = scaler.fit_transform(X_features)

# === Load correct submission for evaluation ===
correct = pd.read_csv("correct_submission.csv")
true_labels = correct['cluster'].values

# === Brute-force hyperparameters ===
n_neighbors_list = [17, 18, 19, 20, 21]
min_dist_list = [0.01, 0.05, 0.1]
n_components_list = [3, 4]
metrics = ['euclidean', 'cosine']
eps_list = [2, 3, 4, 5, 6, 7]
min_samples_list = [5, 10, 15]

best_acc = 0
best_params = None

for n_neighbors, min_dist, n_components, metric in product(
        n_neighbors_list, min_dist_list, n_components_list, metrics):

    # UMAP
    reducer = umap.UMAP(
        n_neighbors=n_neighbors,
        min_dist=min_dist,
        n_components=n_components,
        metric=metric,
        random_state=42
    )
    X_umap = reducer.fit_transform(X_scaled)
    for eps, min_samples in product(
        eps_list, min_samples_list):
        # DBSCAN
        dbscan = DBSCAN(eps=eps, min_samples=min_samples)
        clusters = dbscan.fit_predict(X_umap)

        # handle noise
        noise_mask = clusters != -1
        final_clusters = clusters.copy()
        if np.any(~noise_mask):
            final_clusters[~noise_mask] = np.max(clusters[noise_mask]) + 1
    
        # Hungarian-matched accuracy
        contingency = cluster.contingency_matrix(true_labels, final_clusters)
        row_ind, col_ind = linear_sum_assignment(-contingency)
        acc = contingency[row_ind, col_ind].sum() / true_labels.size
    
        # Update best
        if acc > best_acc:
            best_acc = acc
            best_params = {
                'n_neighbors': n_neighbors,
                'min_dist': min_dist,
                'n_components': n_components,
                'metric': metric,
                'eps': eps,
                'min_samples': min_samples
            }
            print(f"New best acc: {best_acc:.4f} with params: {best_params}")
    if best_acc == 1.0 :
        break
print("\n=== Brute-force finished ===")
print("Best accuracy:", best_acc)
print("Best hyperparameters:", best_params)
````

**1. Dimensionality Reduction**: Which dimensionality reduction methods helped you determine the correct number of clusters? Why did some methods work better than others? Explain the differences between PCA, t-SNE, and UMAP.
I used UMAP for dimensionality reduction. I also tried PCA, but the results weren’t as clear for identifying clusters. UMAP worked better because it’s a non-linear method, so it can capture more complex structures in the data that PCA, which is linear, can’t.
The Differenc between them is: 
PCA: Linear projection, fast, preserves variance, struggles with nonlinear data.  
t-SNE: Nonlinear, preserves local neighborhoods, stochastic, global distances may be distorted.  
UMAP: Nonlinear, balances local and global structure, deterministic with fixed seed, often produces well-separated clusters.  
In my case, UMAP gave the clearest visualization of clusters, making it easier to determine the correct number of clusters.

**3. Data Type Analysis**: What is the hidden data type in the dataset? How does this information influence feature extraction for clustering? Can it explain why some dimensionality reduction algorithms are more effective?
The hidden data type in the dataset looks like to be a continuous numerical telemetry or signal-derived measurements — possibly representing server performance metrics or network activity patterns. This suggests that the data lie on a nonlinear manifold rather than forming linearly separable clusters in the original high-dimensional space.

Because of this, nonlinear dimensionality reduction methods such as UMAP (and to some extent t-SNE) are more effective. PCA, being linear, can only capture variance along straight directions, which fails to preserve the subtle local structure of these nonlinear patterns. UMAP, on the other hand, learns both local and global relationships, making the latent structure (e.g., server groups) far more separable in low-dimensional space. This explains why UMAP produced cleaner, well-defined clusters compared to PCA in this task.

````python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.cluster import KMeans
from scipy.ndimage import center_of_mass

feat_cols = [c for c in data.columns if c.startswith('col_')]
n_feats = len(feat_cols)
print("Number of feature columns:", n_feats)
assert n_feats == 16384, "Expect 16384 columns (128x128 grid)."

# --- reshape to images: rows x 128 x 128 ---
side = 128
X_flat = data[feat_cols].values.astype(float)
X_imgs = X_flat.reshape((-1, side, side))

# --- visualize a few random rows to sanity-check ---
plt.figure(figsize=(10,3))
for i, idx in enumerate(np.random.choice(len(X_imgs), size=5, replace=False)):
    plt.subplot(1,5,i+1)
    plt.imshow(X_imgs[idx], origin='lower')   # origin='lower' for natural coords
    plt.title(f'row {idx}')
    plt.axis('off')
plt.suptitle('Random sample images (128x128)')
plt.show()

# --- compute centroids (center of mass) for each image ---
centroids = np.array([center_of_mass(img) for img in X_imgs])  # returns (y, x)
# convert to (x, y) order and normalize to [0,1] if desired
centroids_xy = np.vstack([centroids[:,1], centroids[:,0]]).T  # columns: x_pixel, y_pixel

# optional: normalize to 0..1
centroids_norm = centroids_xy / (side - 1)

# --- cluster centroids (try k equal to the number of clusters you believe) ---
# if you used k=4 in UMAP earlier, try k=4 here too
k = 4
km = KMeans(n_clusters=k, random_state=42, n_init=20)
cent_labels = km.fit_predict(centroids_norm)
print("Cluster centers (pixel frac coords):", km.cluster_centers_)

# --- quick scatter of centroids colored by cluster ---
plt.figure(figsize=(6,5))
plt.scatter(centroids_xy[:,0], centroids_xy[:,1], c=cent_labels, s=6, cmap='tab10')
plt.gca().invert_yaxis()   # top-left vs bottom-left depending on origin; adjust to taste
plt.title('Image centroids colored by centroid-cluster')
plt.xlabel('x pixel')
plt.ylabel('y pixel')
plt.show()

# --- compute mean image per cluster to find hotspot pixels ---
mean_images = []
for cl in range(k):
    mean_img = X_imgs[cent_labels==cl].mean(axis=0)
    mean_images.append(mean_img)
    plt.figure(figsize=(4,3))
    plt.imshow(mean_img, origin='lower')
    plt.title(f'Mean image, cluster {cl} (n={(cent_labels==cl).sum()})')
    plt.colorbar(shrink=0.6)
    plt.axis('off')
    plt.show()

# --- for each mean image, find top pixel(s) coordinates (hotspots) ---
hotspots = []
for mean_img in mean_images:
    # flatten and pick top 5 pixels
    flat = mean_img.ravel()
    top_idx = np.argsort(flat)[-5:]   # top 5
    ys, xs = np.unravel_index(top_idx, (side, side))
    hotspots.append(list(zip(xs, ys)))
hotspots
````
